```python
from pyspark.sql import functions as F

data1 = [
    (101,1,1200,"COMPLETED"),
    (102,2,800,"COMPLETED"),
    (103,1,500,"CANCELLED"),
    (104,3,2000,"COMPLETED"),
    (105,2,1500,"COMPLETED"),
    (106,1,1000,"COMPLETED"),
    (107,3,1000,"COMPLETED"),
    (108,1,1200,"COMPLETED"),
    (109,2,1500,"COMPLETED"),
    (110,3,1000,"COMPLETED"),
    (111,1,1000,"COMPLETED"),
    (112,2,1000,"COMPLETED")
]

columns = ("order_id","customer_id","amount","status")

orders = spark.createDataFrame(data1,columns)

data2 = [
    (1,"ravi","Hyderabad"),
    (2,"arjun","Chennai"),
    (3,"sita","Hyderabad"),
    (4,"priya","Bangalore"),
    (5,"kiran","Chennai"),
    (6,"anu","Hyderabad"),
    (7,"manoj","Bangalore"),
    (8,"divya","Hyderabad"),
    (9,"rahul","Chennai"),
    (10,"sneha","Hyderabad")
]

columns = ("customer_id","name","city")

customers = spark.createDataFrame(data2,columns)

orders.display()
customers.display()
```

```python
df1 = orders.filter(F.col("status") == "COMPLETED")\
.groupBy("customer_id")\
.agg(F.sum("amount").alias("total_amount"), F.count("order_id").alias("order_count"))

# display(df1)
Join = df1.join(customers, ["customer_id"], how="inner")

df = Join.filter(F.col("total_amount") >= 1500)
# display(df)

df_sel = df.select("customer_id","name","city","total_amount","order_count").orderBy(F.col("total_amount").desc())

display(df_sel)


```
