```python
customers = (spark.read.option("header", True) 

             .schema("""customer_id INT, first_name STRING, last_name STRING, 

                        city STRING, state STRING, signup_date DATE""") 

             .csv(("/Volumes/basicspyspark/default/learn/datasets/customers.csv"))) 
customers.show(10)
```

```python

customers.printSchema()
print(customers.columns)

print(customers.dtypes)

print(customers.count())

customers.select("first_name", "city").show()

customers.limit(3).show()


```

```python
customers.groupBy("city").count().show()
customers.groupBy("city").avg("customer_id").show()
customers.groupBy("city").sum("customer_id").show()
customers.groupBy("city").max("customer_id").show()
customers.groupBy("city").min("customer_id").show()
# customers.groupBy("city").agg("customer_id").show()
```
