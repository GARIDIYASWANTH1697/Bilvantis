```python
from pyspark.sql import functions as F

numbers = spark.createDataFrame([(1,), (2,), (3,), (4,), (5,), (6,)], ["num"])

# 1. map() — multiply each value by 2
print(numbers.select((F.col("num") * 2).alias("doubled")).collect())

# 2. filter() — even numbers only
print(numbers.filter(F.col("num") % 2 == 0).collect())

# 3. distinct()
rdd2 = spark.createDataFrame([(1,), (2,), (2,), (3,), (3,), (4,)], ["num"])
print(rdd2.distinct().collect())

# 4. sortBy() — descending order
print(numbers.orderBy(F.col("num").desc()).collect())
```

```python
from pyspark.sql import functions as F

numbers = spark.createDataFrame([(10,), (20,), (30,), (40,), (50,)], ["num"])

print("Count:", numbers.count())
print("First:", numbers.first())
print("Take:", numbers.take(3))
print("Sum:", numbers.agg(F.sum("num")).collect()[0][0])
print("Maximum:", numbers.agg(F.max("num")).collect()[0][0])
```
