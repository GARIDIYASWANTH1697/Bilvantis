# Spark Joins: Inner, Left, Right, Outer, Semi and Anti

A quick hands-on tour of every common PySpark join type using two small sample tables (employees and departments), showing what rows each join keeps and drops.

```python
# Sample data: employees and departments
employees = spark.createDataFrame(
    [(1, "Alice", 101), (2, "Bob", 102), (3, "Carol", None), (4, "Dan", 104), (5, "Eve", 105)],
    "emp_id INT, name STRING, dept_id INT",
)

departments = spark.createDataFrame(
    [(101, "Engineering"), (102, "Marketing"), (104, "Finance"), (106, "Legal")],
    "dept_id INT, dept_name STRING",
)

display(employees)
display(departments)
```

```python
# INNER JOIN: only employees with a matching department
employees.join(departments, "dept_id", "inner").display()
```

```python
# LEFT JOIN: all employees; NULLs where no department matches
employees.join(departments, "dept_id", "left").display()
```

```python
# RIGHT JOIN: all departments; NULLs where no employee belongs to them
employees.join(departments, "dept_id", "right").display()
```

```python
# FULL OUTER JOIN: every employee and every department, matched where possible
employees.join(departments, "dept_id", "outer").display()
```

```python
# LEFT SEMI JOIN: employees that HAVE a matching department (columns from left only)
employees.join(departments, "dept_id", "left_semi").display()
```

```python
# LEFT ANTI JOIN: employees with NO matching department (columns from left only)
employees.join(departments, "dept_id", "left_anti").display()
```

## Quick recap

| Join type | Keeps | Result columns |
| --- | --- | --- |
| `inner` | Only rows with a match on both sides | Left + Right |
| `left` | All left rows, NULLs where unmatched | Left + Right |
| `right` | All right rows, NULLs where unmatched | Left + Right |
| `outer` | All rows from both sides | Left + Right |
| `left_semi` | Left rows that have a match (a filtered left) | Left only |
| `left_anti` | Left rows with NO match (the complement of semi) | Left only |
