```python
from pyspark.sql import SparkSession, functions as F 

  

spark = SparkSession.builder.appName("day3-assignment").getOrCreate() 

  

customers = (spark.read.option("header", True) 

             .schema("""customer_id INT, first_name STRING, last_name STRING, 

                        city STRING, state STRING, signup_date DATE""") 

             .csv(("/Volumes/basicspyspark/default/learn/datasets/customers.csv"))) 

  

orders = (spark.read.option("header", True) 

          .schema("""order_id INT, order_date DATE, customer_id INT, 

                     order_status STRING, amount DOUBLE""") 

          .csv(("/Volumes/basicspyspark/default/learn/datasets/orders.csv"))) 

  

products = (spark.read.option("header", True) 

            .schema("""product_id INT, product_name STRING, 

                       category STRING, price DOUBLE""") 

            .csv("/Volumes/basicspyspark/default/learn/datasets/products.csv")) 
```

```python
joined = customers.join(orders, "customer_id", "inner")
joined.show()
```

```python
from pyspark.sql import functions as F

# 1. withColumn: add one column at a time
#    full_name = first_name + last_name, amount_with_tax = amount * 1.1
enriched = (joined
            .withColumn("full_name", F.concat_ws(" ", "first_name", "last_name"))
            .withColumn("amount_with_tax", F.round(F.col("amount") * 1.1, 2)))

enriched.select("full_name", "order_id", "amount", "amount_with_tax").show(10)

# 2. Conditional column with when/otherwise
#    size_bucket based on order amount
with_bucket = enriched.withColumn(
    "size_bucket",
    F.when(F.col("amount") >= 500, "HIGH")
     .when(F.col("amount") >= 100, "MEDIUM")
     .otherwise("LOW")
)

with_bucket.select("full_name", "amount", "size_bucket").show(10)

# 3. Add multiple columns at once using select / withColumns
#    - select: build a whole new projection
#    - withColumns (plural): add several columns in a single call
with_multi = enriched.withColumns({
    "order_year": F.year("order_date"),
    "is_closed": F.col("order_status") == "CLOSED"
})

with_multi.select("full_name", "order_date", "order_year", "is_closed").show(10)

# 4. selectExpr: SQL-style expressions to derive new columns
joined.selectExpr(
    "first_name",
    "last_name",
    "concat(first_name, ' ', last_name) AS name_full",
    "round(amount * 1.1, 2) AS amount_with_tax",
    "year(order_date) AS order_year"
).show(10)
```

# Run SQL queries on joined data

Any DataFrame can be queried with SQL in two steps:

1. **Register it as a temporary view** with `df.createOrReplaceTempView("name")` — this makes the DataFrame visible as a SQL table for the current Spark session.
2. **Query it with `spark.sql("SELECT ...")`** — the result is itself a DataFrame, so you can chain `.show()`, `.filter()`, etc. on it.

Everything you did with DataFrame methods has a SQL equivalent:

| DataFrame method | SQL equivalent |
| --- | --- |
| `select(...)` | `SELECT ...` |
| `filter(...)` | `WHERE ...` |
| `groupBy(...).agg(...)` | `GROUP BY ...` with aggregates |
| `withColumn(...)` | expression in SELECT with alias |
| `orderBy(...)` | `ORDER BY ...` |

You can also use native `%sql` notebook cells directly — on this compute, `%sql --name my_result` stores the output in a Python variable you can reuse.

```python
# Register the joined DataFrame as a temporary view — then it becomes a SQL table
joined.createOrReplaceTempView("customer_orders")

# Same analysis as before, but in SQL instead of DataFrame methods
spark.sql("""
    SELECT city,
           COUNT(*)          AS total_orders,
           ROUND(AVG(amount), 2) AS avg_amount,
           MAX(amount)       AS max_amount
    FROM customer_orders
    GROUP BY city
    ORDER BY total_orders DESC
""").show()

# SQL WHERE filter (equivalent of joined.filter(...))
spark.sql("""
    SELECT first_name, last_name, order_id, amount
    FROM customer_orders
    WHERE order_status = 'CLOSED' AND amount > 5000
    ORDER BY amount DESC
""").show(10)

# Derived columns in SQL (equivalent of withColumn / selectExpr)
spark.sql("""
    SELECT CONCAT(first_name, ' ', last_name) AS full_name,
           YEAR(order_date)                   AS order_year,
           ROUND(amount * 1.1, 2)             AS amount_with_tax
    FROM customer_orders
    LIMIT 10
""").show()
```
