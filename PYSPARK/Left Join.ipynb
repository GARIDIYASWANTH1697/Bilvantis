```python
from pyspark.sql import functions as F

customers = spark.createDataFrame([
    (1, "Ravi"),
    (2, "Sita"),
    (3, "Arjun"),
    (4, "Priya"),
    (5, "Kiran"),
    (6, "Divya"),
    (7, "Rahul"),
    (8, "Sneha"),
    (9, "Manoj"),
    (10, "Anu")
], ["customer_id", "name"])

orders = spark.createDataFrame([
    (101, 1, 500),
    (102, 3, 700),
    (103, 3, 200),
    (104, 5, 900),
    (105, 7, 400),
    (106, 9, 600),
    (107, 11, 300),
    (108, 12, 800),
    (109, 13, 250),
    (110, 15, 450)
], ["order_id", "customer_id", "amount"])
```

```python
result = customers.join(orders, "customer_id", "left_semi")
result.show()

left_anti = customers.join(orders, "customer_id", "left_anti")
left_anti.show()
```

```python
c = customers.alias("c")
o = orders.alias("o")

result = c.join(
    o,
    F.col("c.customer_id") == F.col("o.customer_id"),
    "left"
).select(
    F.col("c.name").alias("customer_name"),
    F.col("o.amount").alias("order_amount")  # orders has no "name" column; using "amount"
)
```

```python
data = [
    (101, "Ravi", "IT", 50000),
    (102, "Sita", "HR", None),
    (103, "Arjun", None, 60000),
    (104, "Priya", "IT", None),
    (105, "Kiran", None, None)
]

columns = ["emp_id", "name", "department", "salary"]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
# df.filter(F.col("salary").isNull()).show()

from pyspark.sql import functions as F

df.filter(F.col("salary").isNull()).show()

df.filter(F.col("salary").isNotNull()).show()

df.select([
    F.count(
        F.when(F.col(c).isNull(), 1)
    ).alias(c)
    for c in df.columns
]).show()


df.dropna().show()

df.show()

df.dropna(how="all").show()
```
