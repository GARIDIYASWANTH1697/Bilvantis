```
%%sql -r dataframe_1
CREATE WAREHOUSE IF NOT EXISTS my_wh 
WITH WAREHOUSE_SIZE = 'XSMALL' 
AUTO_SUSPEND = 60 
AUTO_RESUME = TRUE;

CREATE DATABASE IF NOT EXISTS practice_db;
CREATE SCHEMA IF NOT EXISTS practice_db.practice_schema;
```

```
%%sql -r dataframe_2
CREATE TABLE IF NOT EXISTS practice_db.practice_schema.sales (
    id INT,
    product STRING,
    amount FLOAT,
    sale_date DATE
);

INSERT INTO practice_db.practice_schema.sales VALUES
(1, 'Laptop', 1200, '2026-11-01'),
(2, 'Phone', 800, '2026-11-02');

SELECT * FROM practice_db.practice_schema.sales;
```
